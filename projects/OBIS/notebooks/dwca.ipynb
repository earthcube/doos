```python
import requests
import io
from dwca.read import DwCAReader
```

```python
# download zip file to a variable
def download_and_load(url):
    # Send a HTTP request to the URL
    response = requests.get(url)
    # Check if the request was successful
    if response.status_code == 200:
        # Load the data into a BytesIO object
        data = io.BytesIO(response.content)
        return data
    else:
        return "Downloading failed with status code {}".format(response.status_code)
```

```python
url = 'https://ipt.vliz.be/eurobis/archive.do?r=smhi-zoobenthos-reg'
data = download_and_load(url)
```

```python
zip_stream = io.BytesIO(data)

with DwCAReader(zip_stream) as dwca:
    print("Core data file is: {}".format(dwca.descriptor.core.file_location)) # => 'occurrence.txt'
    core_df = dwca.pd_read('event.txt', parse_dates=True)
```
